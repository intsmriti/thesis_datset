```python
# ==========================================
# HEALTH IT DATASET — BASIC AUDIT
# ==========================================

health_it = pd.read_csv(
    "hospital-promoting-interoperability-chpl-linkage.csv"
)

print("Shape:", health_it.shape)

print("\nColumns:")
print(health_it.columns.tolist())

print("\nFirst 5 rows:")
display(health_it.head())

print("\nYears:")
print(
    health_it["year"]
    .value_counts(dropna=False)
    .sort_index()
)

print("\nUnique facilities:")
print(health_it["Facility.ID"].nunique())

print("\nUnique facility names:")
print(health_it["Facility.Name"].nunique())
```

```python
# ==========================================
# ROWS PER FACILITY
# ==========================================

rows_per_facility = (
    health_it
    .groupby("Facility.ID")
    .size()
)

print("Rows per facility:")
print(rows_per_facility.describe())

print("\nFacilities with more than 1 row:",
      (rows_per_facility > 1).sum())

print("\nFacilities with exactly 1 row:",
      (rows_per_facility == 1).sum())
```

```python
# ==========================================
# WHY ARE THERE MULTIPLE ROWS PER FACILITY?
# ==========================================

# Select a facility that has many records
example_facility = (
    health_it["Facility.ID"]
    .value_counts()
    .idxmax()
)

print("Example Facility.ID:", example_facility)

example = health_it[
    health_it["Facility.ID"] == example_facility
].copy()

print("\nNumber of rows:", len(example))

display(example)
```

```python
for col in health_it.columns:
    print(
        f"{col}: "
        f"{example[col].nunique(dropna=False)} unique values"
    )
```

```python
# ==========================================
# HEALTH IT: YEAR AND CRITERION AUDIT
# ==========================================

print("Year distribution:")
print(health_it["year"].value_counts(dropna=False).sort_index())

print("\nInteroperability criterion distribution:")
print(
    health_it[
        "Meets.criteria.for.promoting.interoperability.of.EHRs"
    ].value_counts(dropna=False)
)

print("\nUnique combinations of facility, year, and criterion:")
print(
    health_it[
        [
            "Facility.ID",
            "year",
            "Meets.criteria.for.promoting.interoperability.of.EHRs"
        ]
    ]
    .drop_duplicates()
    .shape
)
```

```python
# ==========================================
# STEP 2 — HEALTH IT IDENTIFIER AUDIT
# ==========================================

print("Health IT Facility.ID examples:")
print(
    health_it["Facility.ID"]
    .head(10)
    .tolist()
)

print("\nHealth IT Facility.ID data type:")
print(
    health_it["Facility.ID"].dtype
)
```

```python
# ==========================================
# STEP 3 — HEALTH IT RECORD STRUCTURE
# ==========================================

# Number of unique facilities by year
facilities_by_year = (
    health_it
    .groupby("year")["Facility.ID"]
    .nunique()
)

print("Unique facilities by year:")
print(facilities_by_year)


# Unique CHPL records by year
print("\nUnique CHPL IDs by year:")
print(
    health_it
    .groupby("year")["chpl_id"]
    .nunique()
)


# Unique products by year
print("\nUnique products by year:")
print(
    health_it
    .groupby("year")["product_database_id"]
    .nunique()
)
```

```python
# ==========================================
# STEP 4 — CHECK FACILITY/YEAR DUPLICATION
# ==========================================

facility_year = (
    health_it[
        ["Facility.ID", "year"]
    ]
    .drop_duplicates()
)

print(
    "Unique facility-year combinations:",
    len(facility_year)
)

print("\nExample facilities with records in both years:")

year_counts = (
    facility_year
    .groupby("Facility.ID")["year"]
    .nunique()
)

print(
    year_counts.value_counts()
)
```
