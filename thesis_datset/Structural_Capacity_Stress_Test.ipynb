# Structural Regional Capacity Stress Test

**Scope:** Exploratory, scenario-based model using actual 2016–2021 CMS Cost Report bed counts and real THREAT hospital-attack observations. **Not** a geocoded accessibility network, real observed occupancy, measured diversion, or validated severity predictor.

**Assumptions:** Numeric metropolitan CBSA codes 10000–89999 represent coarse regions; bed counts 1–2000 are retained under a prespecified plausibility screen; all beds of alternative same-CBSA hospitals are assumed potentially substitutable; occupancy 70/85/95% and attacked-hospital capacity loss 10/25/50% are hypothetical. Real-world service specialization, travel time, bed staffing, diversion policies and baseline demand are unmodeled. Results must not be described as actual hospital spare capacity or observed patient shortfalls.

**Equation:** displaced bed equivalents = attacked hospital beds × assumed capacity-loss fraction; hypothetical spare regional beds = alternative reported beds × (1 − assumed occupancy); unmet bed equivalents = max(0, displaced − hypothetical spare).

**Data quality:** repeated CCN-year reports are resolved by valid beds then latest fiscal end date; results depend on this rule. Excluded nonmetropolitan/invalid CBSA records are not assigned fabricated regions. No claims of historical geospatial precision.
